## Edge probabilities and conditional probabilities

We plot a sample edge-probability plot for a small system size, using the floating point Kasteleyn matrix inverse.
The corresponding figure in the paper with a much larger system size is generated using K^{-1} calculated using fraction-free LU decomposition.

```python
mpl.rcParams['figure.dpi'] = 80
print('Sample: small size plot using floating point inverse')
k=kasteleyn_shape(60,octagon_fortress_coord,sparse=False)
kinv=np.linalg.inv(k)
e=edge_prob_plot(60,octagon_fortress_coord,k,kinv,curve=True)
```

We can compute conditional probabilities of certain dimer patterns. The following code may take some time and RAM to invert the Kasteleyn matrix

```python
mpl.rcParams['figure.dpi'] = 30 
k=kasteleyn_shape(100,octagon_square_coord, sparse=False); kinv=np.linalg.inv(k)
conditional_prob_plot(100,octagon_square_coord,k,kinv,[[(50,100),(50,101)]],5,savename='',colorbar=False,pt=30,edgewidth=15);
conditional_prob_plot(100,octagon_square_coord,k,kinv,[[(50,100),(50,101)],[(51,100),(51,101)]],5,savename='',pt=30,edgewidth=15);
```

```python
mpl.rcParams['figure.dpi'] = 80
k=kasteleyn_shape(80,octagon_fortress_coord, sparse=False); kinv=np.linalg.inv(k)
conditional_prob_plot(80,octagon_fortress_coord,k,kinv,[[(58,88),(58,89)]],78,inset=True,edgewidth=2,insetedgewidth=4.5,inset_radius=7,savename='',bdybox=False,ocurve=True);
```
